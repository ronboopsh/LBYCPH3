**A. Multi-layered Artificial Neural Networks**

&emsp;&emsp;&emsp;Multilayer neural networks contain more than one computational layer.  The additional intermediate layers (between input and output) are referred to as hidden layers because the computations performed are not visible to the user. The specific architecture of multilayer neural networks is referred to as feed-forward networks, because successive layers feed into one another in the forward direction from input to output.

The feedforward network architecture is defined by the following:
1. The number of input features $n$
2. The number of outputs $m$
3. The number of layers $L$
4. The number of neurons per layer $n_1,n_2,...,n_{L-1}$
5. The activation function of neurons (e.g., sigmoid)

The number of weights and biases is derived from the number of layers and the number of neurons per layer. The architecture is illustrated in Figure 2.1 highlighting some interconnecting weights between the layer nodes.

<center><img src="https://drive.google.com/uc?export=view&id=1po3s9FI2Dg3E4dTYyrP_d6YqpjNHFuYJ"><br/>Figure 2.1. The feedforward neural network architecture</center>

&emsp;&emsp;&emsp;The activation of the neurons at layer $k$ in vectorized form can be expressed as:
$$
\textbf{a}^\left[k\right]=\sigma\left(\textbf{W}^\left[k\right]\textbf{a}^\left[k-1\right]+\textbf{b}^\left[k\right]\right)
$$
where $\textbf{a}^\left[k\right]$ is the vector of neuron activations with shape $\left(n_k\times 1\right)$, $\textbf{b}^\left[k\right]$ is the vector of biases with shape $\left(n_k\times 1\right)$, $\textbf{W}^\left[k\right]$ is the matrix of weights connecting the neurons from layer $k-1$ to $k$ with shape $\left(n_k\times n_{k-1}\right)$ and $\sigma$ is the sigmoid function.

&emsp;&emsp;&emsp;The output of the neural network is determined layer by layer. The activations of the first layer is calculated from the inputs. The activations of the second layer is calculated once the activation of the first layer is known. The third layer activations is computed once the activation of the second layer is known, and so on up to the output layer. The activations of the output layer ($\textbf{a}^\left[L-1\right]$) is the output of the network ($\hat{\textbf{y}}$).
$$
\textbf{a}^\left[1\right]=\sigma\left(\textbf{W}^\left[1\right]\textbf{x}+\textbf{b}^\left[1\right]\right) \\
\textbf{a}^\left[2\right]=\sigma\left(\textbf{W}^\left[2\right]\textbf{a}^\left[1\right]+\textbf{b}^\left[2\right]\right) \\
\vdots\\
\hat{\textbf{y}}=\textbf{a}^\left[L\right]=\sigma\left(\textbf{W}^\left[L\right]\textbf{a}^\left[L-1\right]+\textbf{b}^\left[L\right]\right)
$$

It focuses on high performance and distributed training. However, it is not as widely used as TensorFlow or PyTorch, has smaller community and fewer resources available.
5. **Microsoft Cognitive Toolkit (CNTK)**: Developed by Microsoft, this framework excels in handling large datasets and computationally intensive tasks due to its optimized architecture and support for various hardware. However, it has a steeper learning curve compared to other frameworks due to less intuitive syntax and less extensive documentation.

3. We will build a class named `FeedForwardNet` which would consists of several methods. The details for each method are as follows:

  (a) **Initialization** (`__init__`): This function is called upon instantiation of the network. It requires a list of sizes of each layer as well as the learning rate for training later. The first element of the list is the number of the input nodes $n$ and the last element is the number of output nodes $m$. Furthermore, this function also initializes weights and biases randomly.

  (b) **Prediction** (`predict`): This function is called to compute the output of the network. It requires an input `X` to be a column vector of features with shape of $\left(n\times 1\right)$. If the input has $p$ data points, then `X` would also have to have $p$ columns. The function returns the output of the network with the dimensions of $\left(m\times p\right)$. It should implement the following across the hidden and output layers:
  $$
  \textbf{a}^\left[1\right]=\sigma\left(\textbf{W}^\left[1\right]\textbf{x}+\textbf{b}^\left[1\right]\right) \\
  \textbf{a}^\left[2\right]=\sigma\left(\textbf{W}^\left[2\right]\textbf{a}^\left[1\right]+\textbf{b}^\left[2\right]\right) \\
  \vdots\\
  \hat{\textbf{y}}=\textbf{a}^\left[L\right]=\sigma\left(\textbf{W}^\left[L\right]\textbf{a}^\left[L-1\right]+\textbf{b}^\left[L\right]\right)
  $$

  (c) **Training** (`fit`): This function is called to train the network. It requires the training inputs `X`, training outputs `y`, and optionally the number of epochs. By default, the number of epochs is set to 100. It performs training using the backpropagation algorithm. The function returns a list of training loss per epoch.

  (d) **Evaluation** (`evaluate`): This function is called to evaluate the model on a test data. It requires the test inputs `X` and test outputs `y`. The function returns the accuracy of the model computed as follows:
  $$
  \text{Accuracy} = \dfrac{1}{p}\sum\limits_{\textbf{x}\in\textbf{X}}\left[\text{argmax}\left(\textbf{y}\right) == \text{argmax}\left(\hat{\textbf{y}}\right)\right]
  $$ where $p$ is the number of test data points.

````python
# Define the sigmoid function
def sigmoid(X):
  return 1.0/(1.0 + np.exp(-X))

# Define the FeedForwardNet class
class FeedForwardNet:
  def __init__(self, layer_sizes, learning_rate):
    self.layer_sizes = layer_sizes
    # Initialize the weights and biases
    self.weights = [np.random.randn(y, x)
                    for x, y in zip(layer_sizes[:-1], layer_sizes[1:])]
    self.biases = [np.random.randn(y, 1) for y in layer_sizes[1:]]
    self.learning_rate = learning_rate

  def predict(self, X):
    # Stating with the inputs, compute the activations of the hidden layers
    # up to the output layer. This can be implemented using for loops.
    # You must implement the activation formulation a = sigmoid(WX + b) for
    # each iteration
    ### --YOUR CODE HERE-- ###

  def fit(self, X, y, epochs=100):
    losses = list() # Create a list to contain the losses per epoch

    for ep in range(epochs):
      print(f"Epoch {ep + 1}...", end="")

      # Initialize gradients for weights and biases
      delta_W = [np.zeros(w.shape) for w in self.weights]
      delta_b = [np.zeros(b.shape) for b in self.biases]

      # Iterate through each training data point
      for xi, yo in zip(X, y):
        # Run the backpropagation algorithm
        ddelta_W, ddelta_b = self.backpropagation(xi.reshape(-1,1),
                                                  yo.reshape(-1,1))
        # Update the gradient weights and biases for each datapoint
        delta_W = [dw + ddw for dw, ddw in zip(delta_W, ddelta_W)]
        delta_b = [db + ddb for db, ddb in zip(delta_b, ddelta_b)]

      # Update the weights and biases for the current epoch
      self.weights = [w - (self.learning_rate/X.shape[0])*dw
                      for w, dw in zip(self.weights, delta_W)]
      self.biases = [b - (self.learning_rate/X.shape[0])*db
                     for b, db in zip(self.biases, delta_b)]

      # Calculate the training loss
      ypred = self.predict(X.T)
      error = np.linalg.norm(y.T - ypred, axis=0)
      cost = np.sum(np.square(error))/y.shape[0]
      losses.append(cost)
      print(f"loss: {cost:.3f}")

    return losses

  def evaluate(self, X, y):
    # Calculate the output of the network and obtain the index with
    # the highest activation using the NumPy argmax function. The network
    # is considered to have correctly predicted the output if it matches with
    # the target. The sum of all the matches divided by the number of the
    # datapoints is the accuracy of the network. You may also use the NumPy
    # where and sum functions.
    ### --YOUR CODE HERE-- ###

  def backpropagation(self, X, y):
    # Initialize incremental gradients for weights and biases
    delta_W = [np.zeros(w.shape) for w in self.weights]
    delta_b = [np.zeros(b.shape) for b in self.biases]

    # forward pass
    activation = X
    activations = [X]
    zvalues = []

    # compute all activations at each layer
    for W, b in zip(self.weights, self.biases):
        z = np.dot(W, activation) + b
        zvalues.append(z)
        activation = sigmoid(z)
        activations.append(activation)

    # backward pass
    delta = sigmoid(zvalues[-1])*(1-sigmoid(zvalues[-1]))*(activations[-1]-y)
    delta_W[-1] = np.dot(delta, activations[-2].T)
    delta_b[-1] = delta

    # compute all gradients at each layer
    for layer in range(2, len(self.layer_sizes)):
      z = zvalues[-layer]
      delta = sigmoid(z)*(1-sigmoid(z))*np.dot(self.weights[-layer+1].T, delta)
      delta_W[-layer] = np.dot(delta, activations[-layer-1].T)
      delta_b[-layer] = delta

    return (delta_W, delta_b)
````
